```python
from pathlib import Path

import pandas as pd

pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 120)


def find_project_root(start: Path) -> Path:
    """현재 위치에서 위로 올라가며 data/raw 폴더가 있는 프로젝트 루트를 찾습니다."""
    start = start.resolve()
    for candidate in [start, *start.parents]:
        if (candidate / "data" / "raw").exists() and (candidate / "book").exists():
            return candidate
    raise FileNotFoundError("프로젝트 루트를 찾지 못했습니다. 노트북을 저장소 안에서 실행해 주세요.")


PROJECT_ROOT = find_project_root(Path.cwd())
DATA_DIR = PROJECT_ROOT / "data" / "raw"

print("프로젝트 루트:", PROJECT_ROOT)
print("데이터 폴더:", DATA_DIR)
```

### 나의 해석과 판단
분석 전 가장 주의해서 확인해야 할 컬럼은 `signup_date`와 `order_date`라고 생각했다. 날짜를 나타내는 값으로 추정되는데, 숫자가 아닌 문자열 타입으로 현재 저장되어 있는 것으로 보인다. 따라서 기간을 설정하여 분석하기 전에 실제 날짜 타입으로 정상적으로 변환될 수 있는지 확인해야 한다.
이 외 age, price, quantity, unit_price 등 이후 숫자 크기별로 정렬하거나 계산할 때 사용해야 하는 값은 숫자형으로 지정되어 있어 추후 분석 시 편리하게 사용할 수 있을 것이라고 생각했다. 다만 `id` 항목들도 숫자형으로 지정되어 있는데, 이는 식별하기 위한 번호이기 때문에 일반적인 숫자처럼 정렬 등으로는 분석하지 않아야 한다고 생각했다. 
orders_items 행이 orders 행보다 많았는데, 이 경우 orders에 여러 아이템이 포함되어 저장되어 있을 수 있다고 생각했다. 따라서 주의하여 확인해야 한다.

### 업무·분석적 의미
데이터 구조를 확인하지 않고 바로 분석을 진행하는 경우, 지금 데이터와 같이 날짜가 문자열 상태면 문자열 상태로 분석하여 기간 분석이 잘못될 수 있다. 또한 ID 등을 일괄적으로 일반적인 숫자형 변수라고 생각해 분석하면 의미 없는 분석이 될 수 있다. 그리고 실제 컬럼명이나 데이터셋 사이 구조를 확인해야 이후 필터링을 하거나 병합을 하는 경우 잘못된 컬럼을 사용하거나 예상하지 못한 결과가 발생하는 것을 막을 수 있다. 따라서 분석 전 데이터 크기나 컬럼명, 데이터 타입, 각 컬럼이 어떤 것을 의미하는지 미리 확인하는 것이 추후 분석 오류를 줄이는 것에 도움을 줄 수 있다. 

### 한계와 추가 확인 사항

### 결과 관찰
4개 CSV의 결측치를 확인한 결과 모든 컬럼에서 결측치 0건, 주요 ID에서 결측치 0건이었다. 즉 결측치가 발견되지 않았다. 중복 여부 경우에도 주요 ID 중복여부를 확인한 결과 모두 중복 0건, 전체 행 중복 역시 모두 0건이었다. 
주요 ID의 중복 여부를 확인한 결과  `customer_id`, `product_id`, `order_id`, `order_item_id` 모두 중복 0건이었다. 전체 행 중복도 모두 0건이었다. 단 `order_items.order_id`에서는 중복이 464건이었다. 

### 나의 해석과 판단
먼저 4개 csv 데이터에서 주요 ID 결측, 주요 ID 중복, 전체 행 중복 모두 나타나지 않았다. 따라서 기본적인 식별을 위해서는 양호한 상태라고 판단했다. `order_items.order_id` 에서는 464건 중복이 나타났으나 잘못된 것이라고 판단하지는 않았다. 한 주문에 포함된 여러 상품을 각각 행으로 기록하는 데이터니까, 하나의 주문에 여러 상품이 포함되어 있는 경우에는 반복될 수 있기 때문이다.
중복 문제가 발견되었을 경우 단순히 중복 건수가 많은 항목부터 삭제하는 게 아니라, 먼저 중복되어서는 안되는 고유한 ID인지, 다른 테이블과 연결하기 위한 키인지 등을 확인해야 한다고 생각한다.
전체 행 중복이 발견된 경우에도 바로 삭제하기보다는 실제 중복 입력인지, 동일한 거래나 이벤트가 여러 번 발생한 정상 기록인지 확인할 필요가 있다. 따라서 결측이나 중복이 발견되더라도 먼저 컬럼의 역할과 데이터 생성 방식을 확인한 뒤 처리 방법을 결정하는 것이 적절하다고 판단했다.
만약 고유해야 하는 주요 ID에서 중복이 발견된다면, 같은 대상을 여러 번 작성한 것인지, 혹은 ID가 잘못된 것인지 등 원인을 파악한 후에, 처리 여부를 결정해야 한다. 

### 업무·분석적 의미
결측치가 있는 경우 필요한 정보가 빠져있어 분석 대상이 누락되거나 집계 결과가 실제 데이터 전체를 반영하지 못할 수 있다. 특히 주요 ID에 문제가 있으면 다른 테이블과 연결하기 어려울 수 있어 반드시 확인이 필요하다.
또한 중복 중에서도 고유해야 하는 주요 ID에 중복이 있으는 경우 데이터를 병합하거나 집계하는 과정에서 똑같은 값이 여러번 계산되어 잘못 분석될 수 있다. 이때 중복이 어떤 유형인지도 파악해야 하는데, 반복될 수 있는 연결 ID를 중복이라는 이유로 삭제하는 경우 한 주문에 포함된 여러 상품 정보가 사라져 데이터가 변형될 수 있다. 
따라서 오류 방지 및 적절한 분석을 위해서 먼저 컬럼 역할을 파악하고, 중복을 판단하는 것이 필요하다.

### 한계와 추가 확인 사항

현재 확인한 범위에서는 키 연결이 적절히 되어있으며 csv 사이 기본적인 연결 관계가 있었다. 따라서 이를 통해 데이터 병합 시 이 ID들을 연결 기준으로 사용할 수 있을 것이라고 생각했다.

    - 금액 관련: 상품 가격과 실제 판매 단가 차이, 할인·배송비·세금 반영 여부 확인
    - 시간 관계: 가입일과 주문일 순서, 분석 기간 및 날짜 누락 여부 확인
    - 분석 단위: 고객 수, 주문 수, 판매량, 매출의 계산 기준과 취소·환불 처리 기준 확인

- 실제 데이터에서 확인한 항목:
    - 키와 관계: 기본키 중복과 테이블 간 외래키 연결이 정상적인지 확인했다. (0건으로 모두 연결되었다)
    - 결측치와 중복: 주요 ID에서 결측과 중복이 모두 0건임을 확인했고, 전체 행 중복 역시 0건임을 확인했다. order_items.order_id 는 중복이 있었으나, 이는 한 주문에 여러 상품이 포함될 수 있어 반복될 수 있는 값임을 확인했다. 
    - 데이터 타입과 값 범위: 범주형 컬럼 값 종류, 빈도를 확인했고 숫자형 컬럼의 범위를 확인했다. 날짜 경우 처음 문자열인 것을 날짜 타입으로 변환 가능한지 확인하였고, 변환 성공했음을 확인했다. 
    - 시간 관계: 주문 날짜의 범위가 2025-07-09부터 2026-07-08까지임을 확인했다. 다만 LLM이 제안한 가입일과 주문일의 선후 관계까지는 아직 확인하지 않았다.



- 채택/수정/보류한 내용:
    - 채택: 각 값 범위 및 빈도 확인, CSV 간 키 관계 확인, 중복, 결측 확인
    - 수정: LLM이 제안안 중복 경우 뭉뚱그려 '중복'이라고만 판단하였는데, 이를 고유해야 하는 ID에 한해서만 중복을 확인하는 것으로 수정한다.  
    - 보류: 
        - 할인, 배송비, 세금 등이 반영되어 있는지 여부 확인
        - 상품 가격과 실제 판매 단가 차이 확인
        - 취소, 환불 처리 기준 및 매출 계산 기준, 판매량 기준 확인
        이 항목들은 현재까지 진행한 점검만으로는 충분히 확인할 수 없었던 정보들이기 때문에 보류하였다.

### 나의 해석과 판단
LLM이 제안한 것 중 가장 유용했던 것은 앞서 `보류`에 작성한 항목들이다. 특히 할인이 들어간 상품 경우 분석 결과에 영향을 미칠 수 있기 때문에 해당 분석 기간 동안 할인이 진행되었는지 등을 분석 시 꼭 고려해야 한다고 생각한다. 따라서 추후 가능하다면 이에 대한 것 역시 추가로 확인해야 한다. LLM이 제안하기 전 생각하지 못했던 항목이었기 때문에 유용하다고 생각했으나 정보가 없어 확인하지는 못했다. 또한 LLM이 제안한 결측치, 중복, 데이터 타입, 숫자형/범주형 데이터 확인, 키 연결 확인 등의 항목은 실습에서도 진행했을 만큼 데이터 분석에서 기본적인 사항이었기 때문에 이를 되짚어 주는 면에서 유용하다고 생각했다.
반면 LLM의 제안을 그대로 받아들이는 것 역시 주의해야 한다고 생각했다. 예를 들어 중복에 대한 내용 경우, 중복되어서는 안되는 컬럼과 중복이 나타날 수 있는 컬럼을 구분해야 하는데 LLM은 그를 구분하지 않았다. 만약 LLM이 중복값을 제거하라고 제안했으면 유의미한 데이터를 지웠을 수도 있으므로 제안을 무조건적으로 수용해서는 안 된다. 


### 한계와 추가 확인 사항
LLM 제안 중 일부는 현재 가진 정보만으로는 확인하기 어려웠다. 예를 들어 할인, 배송비, 세금 반영 여부, 가입일과 주문일의 선후 관계 등은 추가적인 정보가 필요하다. 또한 데이터만으로는 취소나 환불 주문을 매출이나 주문수에 포함할 것인지 결정할 수 없다. 따라서 분석 기준을 어떻게 잡을 것인지 추가적으로 검토가 필요하다. 또한 LLM의 경우 분석 시 고려해야 할 정보 (중복의 유형 등)를 제외하고 언급할 수 있으므로 참고자료로만 사용하고 사람이 판단해야 한다.

```python
'''1. 4개 CSV 파일의 행과 열 개수를 하나의 표로 정리하세요.'''
shape_summary = pd.DataFrame([
    {
        "dataset": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
    }
    for name, df in datasets.items()
])

shape_summary
```

```python
'''2. 각 파일의 결측치 개수 확인하세요.'''
missing_summary = pd.concat(
    [df.isna().sum().rename(name) for name, df in datasets.items()],
    axis=1,
).fillna("").astype(str)

missing_summary
```

```python
'''2. 각 파일의 결측치 비율을 확인하세요.'''
missing_rate_summary = pd.concat(
    [(df.isna().mean() * 100).round(2).rename(name) for name, df in datasets.items()],
    axis=1,
).fillna("")

missing_rate_summary
```

```python
'''3. `orders.order_date`를 날짜 타입으로 변환하고 데이터 기간을 확인하세요.'''
orders["order_date"] = pd.to_datetime(orders["order_date"], errors="coerce")
orders.assign(
    order_year=orders["order_date"].dt.year,
    order_month=orders["order_date"].dt.month,
    order_day_name=orders["order_date"].dt.day_name(),
).head()
```

```python
'''4. `order_items`와 `products`를 병합해 카테고리별 주문 금액을 계산하세요.'''
order_items_with_products = order_items.merge(
    products,
    on="product_id",
    how="left",
)

order_items_with_products["line_amount"] = (
    order_items_with_products["quantity"] * order_items_with_products["unit_price"]
)
category_sales = (
    order_items_with_products
    .groupby("category", as_index=False)["line_amount"]
    .sum()
    .sort_values("line_amount", ascending=False)
)

category_sales
```
