### 나의 해석과 판단

현재 확인된 결과에서 최솟값 및 최댓값만 보았을 때 비정상적으로 보이는 값은 없었다. 모두 양수값이었으며, 최대 최소값의 경우 정수 형태로 나타났다. 다만 값이 범위 안에 있단 사실만으로 정확한 데이터라고 단정하기는 어렵다. 특히 평균 가격이 10만원 이상으로 높아 이 평균값이 실제로 item의 가격을 전체적으로 반영하는 값인지, 혹은 최댓값 등이 영향을 크게 준 것인지 확인해야 한다. outlier 값에 의해 나타난 평균인 경우 중앙값 역시 봐야 한다. 이 데이터 경우 중앙값이 unit_price 기준 111000원, 평균 108561.5원이므로 비교적 전체 데이터를 잘 반영한다고 볼 수 있다. 다만 price와 unit_price의 최소, 최댓값은 같으나 총 count나 mean이 약간 다르므로 그 차이가 어디에서 기인하는 것인지 확인해야 한다.
또한 숫자로 보이는 컬럼이라도 실제 역할을 구분해서 해석해야 한다고 생각했다. `customer_id`, `product_id`, `order_id`, `order_item_id`는 `int64`로 저장되어 있으나 식별을 위한 값이기 때문에 평균, 크기 등을 분석하는 것은 의미가 없닥다고 생각했다. 또한 숫자가 문자열로 저장되는 경우도 있을 수 있으므로 데이터 타입을 먼저 확인해야 한다. 다만 이번 실습에서는 이전 단계에서 해당 열의 타입을 모두 확인했고, 날짜 제외 다른 숫자형으로 되어있어야 할 항목은 int64 형태임을 확인했다.
또한 범주형 데이터 (order_status) 의 경우 빈도를 살펴봐야 하는데, 빈도가 특정값이 높거나 낮다는 이유만으로 이상값이라고 생각해서는 안된다. completed된 주문이 많은데, 이것이 정상적인 결과인지 혹은 데이터 수집 과정에서 편향되게 수집된 것인지 현재 결과로 파악이 어려우므로 추가적인 조사가 필요하다. 어떻게 이 데이터를 받았는지, 어떤 필터나 제약 조건 하에서 수집한 결과인지 살펴봐야 한다. 

### 업무·분석적 의미
숫자형 데이터의 범위와 분포를 확인해야 하는데, 이를 확인하지 않고 분석하는 경우 입력 오류나 비정상적인 값이 통계값(평균 등)에 영향을 줘 잘못된 해석을 유도할 수 있다. 따라서 분석 전에 최댓값, 최소값 등을 확인하고 범위를 확인해 정상적인 값인지 검토해야 한다. 범주형 데이터 경우에도 빈도를 미리 확인해 특정 값에 데이터가 지나치게 집중되어 있지는 않은지, 예상하지 못한 범주도 존재하는지 확인해야 한다. 이를 확인하지 않고 비교를 진행하면 표본 수 차이를 고려하지 않은 채 결과를 해석하게 될 수도 있다. 또한 날짜 컬럼 경우 미리 문자열 상태에서 날짜 타입으로 변환해야 기간별 분석이 가능하다. 

### 한계와 추가 확인 사항
현재 숫자형 데이터에서 본 기본적인 사항으로는 실제로 개별 값이 실제 업무적으로 정상적인 값인지 판단하기 어렵다. 특히 상품별로 적정가가 존재하는데, 이 값에 해당하는 값인지는 실제 데이터 값을 확인해야 한다. 또한 한 주문에서 가능한 최대 수량이나 최대값 등 기준이 있는 경우 이 역시 확인해야 한다. 
또한 범주형 데이터에서 분포가 두드러지게 나타나는 경우 그것이 데이터 수집 과정에서 발생한 편향의 문제인지, 실제로 나타나는 분포인지 확인해야 한다. 
price와 unit_price의 최소, 최댓값은 같으나 총 count나 mean이 약간 다르므로 그 차이가 어디에서 기인하는 것인지 확인해야 한다.

## 4. CSV 간 키 관계 검증 
- 없는 `customer_id`: 없음 (`orders.customer_id → customers.customer_id`)
- 없는 `order_id`: 없음 (`order_items.order_id → orders.order_id`)
- 없는 `product_id`: 없음 (`order_items.product_id → products.product_id`)

![PK FK 관계 검증](images/step04_relationship.png)

### 결과 관찰
csv 사이 키 관계를 확인한 결과, `orders.customer_id → customers.customer_id`, `order_items.order_id → orders.order_id`, `order_items.product_id → products.product_id` 에 존재하지 않는 값 모두 0이었다. 
키 관계를 확인한 후 `order_items`와 `products`를 `product_id` 기준으로 병합한 결과 상품명, 카테고리 등 상품 정보가 주문 상세 데이터에 연결되었다. 이후 quantity * unit_price를 통해 각 행의 line_amount를 계산하고 카테고리 별로 집계할 수 있었다.

### 나의 해석과 판단
csv 사이 키 관계를 확인한 결과, `orders.customer_id → customers.customer_id`, `order_items.order_id → orders.order_id`, `order_items.product_id → products.product_id` 에 존재하지 않는 값 모두 0이었으므로 현재 데이터에서 주문 데이터의 고객 ID, 주문 상세 데이터의 주문 ID, 상품 ID가 각각 연결 대상인 부모 테이블에 모두 존재하는 것이라고 해석할 수 있었다. 
현재 확인한 범위에서는 키 연결이 적절히 되어있으며 csv 사이 기본적인 연결 관계가 있었다. 따라서 이를 통해 데이터 병합 시 이 ID들을 연결 기준으로 사용할 수 있을 것이라고 생각했다.
다만 연결되지 않은 ID가 있다고 해서 해당 행을 바로 삭제해서는 안 된다고 생각했다. 부모 테이블 데이터가 누락되었거나, 두 파일의 ID의 데이터 타입이 다르거나, 문자열의 앞 뒤 공백이 다른 등 형식의 작은 차이가 있을 수 있다. 혹은 파일의 수집 시점이 달라서 한쪽에만 데이터가 존재하는 경우가 있을 수도 있다. 따라서 키 관계에 문제가 있는 경우, 단순히 문제가 있는 행을 제거하기보다는 어떤 테이블에서 문제가 발생했는지 확인하고, 그 원인을 확인한 뒤에 처리 방법을 결정해야 한다고 생각했다. 

### 업무·분석적 의미
csv 사이 키 관계를 확인하는 것은 추후 데이터 병합 과정 등에서 필요한 과정이다. 정상적으로 키 관계가 연결되어 있지 않은 경우 병합하는 과정에서 주문이나 상품 정보가 일부 누락될 수도 있고, 키가 중복되어있거나 잘못 연결된 경우에는 하나 행이 여러 행으로 계산되어 실제보다 크게 반영될 수도 있다. 따라서 csv 분석 전 사용할 ID가 서로 잘 대응하는지 확인하는 것이 중요하다.

관계에서 연결되지 않는 ID가 없었기 때문에 상품 정보를 병합하고 `line_amount`를 계산하는 과정까지 정상적으로 진행할 수 있었다. 다만 이 결과는 본격적인 EDA 결과라기보다 데이터가 실제로 연결 가능한지를 확인하기 위한 기본 점검으로 보는 것이 적절하다.

### 한계와 추가 확인 사항
이번 키 관계를 확인하는 과정을 통해 세 관계 모두 정상적으로 연결되어있다고 판단할 수 있었으나, 이것만으로 데이터의 전체적인 정확성이 보장된다고 보기에는 어렵다. 또한 ID가 정상적으로 연결된다고 해도, 실제 고객 - 주문 - 상품 등이 바르게 대응되어 있는지는 확인할 수 없다. 이번 실습은 csv 사이 연결 구조가 제대로 유지되고 있는지만 파악한 것이라고 볼 수 있다. 따라서 구체적인 분석 기준이나 값 자체가 업무적으로 정확한지는 추후 확인해야 한다.

    - 결측치와 중복: NULL, 빈값, 중복 행이 있는지 확인
    - 데이터 타입과 값 범위: 숫자·날짜 형식이 올바르고 비정상 값이 없는지 확인
    - 범주형 값: 범주값의 오탈자, 공백, 대소문자 차이로 값이 나뉘지 않았는지 확인
    - 금액 관련: 상품 가격과 실제 판매 단가 차이, 할인·배송비·세금 반영 여부 확인
    - 시간 관계: 가입일과 주문일 순서, 분석 기간 및 날짜 누락 여부 확인
    - 분석 단위: 고객 수, 주문 수, 판매량, 매출의 계산 기준과 취소·환불 처리 기준 확인

- 실제 데이터에서 확인한 항목:
    - 키와 관계: 기본키 중복과 테이블 간 외래키 연결이 정상적인지 확인했다. (0건으로 모두 연결되었다)
    - 결측치와 중복: 주요 ID에서 결측과 중복이 모두 0건임을 확인했고, 전체 행 중복 역시 0건임을 확인했다. order_times.order_id 는 중복이 있었으나, 이는 한 주문에 여러 상품이 포함될 수 있어 반복될 수 있는 값임을 확인했다. 
    - 데이터 타입과 값 범위: 범주형 컬럼 값 종류, 빈도를 확인했고 숫자형 컬럼의 범위를 확인했다. 날짜 경우 처음 문자열인 것을 날짜 타입으로 변환 가능한지 확인하였고, 변환 성공했음을 확인했다. 
    - 시간 관계: 날짜 범위를 확인했다. 




- 채택/수정/보류한 내용:
    - 채택: 각 값 범위 및 빈도 확인, CSV 간 키 관계 확인, 중복, 결측 확인
    - 수정: LLM이 제안안 중복 경우 뭉뚱그려 '중복'이라고만 판단하였는데, 이를 고유해야 하는 ID에 한해서만 중복을 확인하는 것으로 수정한다.  
    - 보류: 
        - 할인, 배송비, 세금 등이 반영되어 있는지 여부 확인
        - 상품 가격과 실제 판매 단가 차이 확인
        - 취소, 환불 처리 기준 및 매출 계산 기준, 판매량 기준 확인
        이 항목들은 현재까지 진행한 점검만으로는 충분히 확인할 수 없었던 정보들이기 때문에 보류하였다.

### 나의 해석과 판단
LLM이 제안한 것 중 가장 유용했던 것은 앞서 `보류`에 작성한 항목들이다. 특히 할인이 들어간 상품 경우 분석 결과에 영향을 미칠 수 있기 때문에 해당 분석 기간 동안 할인이 진행되었는지 등을 분석 시 꼭 고려해야 한다고 생각한다. 따라서 추후 가능하다면 이에 대한 것 역시 추가로 확인해야 한다. 다만 LLM이 제안하기 전 생각하지 못했던 항목이었기 때문에 유용하다고 생각했다. 또한 LLM이 제안한 결측치, 중복, 데이터 타입, 숫자형/범주형 데이터 확인, 키 연결 확인 등의 항목은 실습에서도 진행했을 만큼 데이터 분석에서 기본적인 사항이었기 때문에 이를 되짚어 주는 면에서 유용하다고 생각했다.
반면 LLM의 제안을 그대로 받아들이는 것 역시 주의해야 한다고 생각했다. 예를 들어 중복에 대한 내용 경우, 중복되어서는 안되는 컬럼과 중복이 나타날 수 있는 컬럼을 구분해야 하는데 LLM은 그를 구분하지 않았다. 만약 LLM이 중복값을 제거하라고 제안했으면 유의미한 데이터를 지웠을 수도 있으므로 제안을 무조건적으로 수용해서는 안 된다. 


### 한계와 추가 확인 사항
LLM 제안 중 일부는 현재 가진 정보만으로는 확인하기 어려웠다. 예를 들어 할인, 배송비, 세금 반영 여부, 가입일과 주문일의 선후 관계 등은 추가적인 정보가 필요하다. 또한 데이터만으로는 취소나 환불 주문을 매출이나 주문수에 포함할 것인지 결정할 수 없다. 따라서 분석 기준을 어떻게 잡을 것인지 추가적으로 검토가 필요하다. 또한 LLM의 경우 분석 시 고려해야 할 정보 (중복의 유형 등)를 제외하고 언급할 수 있으므로 참고자료로만 사용하고 사람이 판단해야 한다.

```python
'''1. 4개 CSV 파일의 행과 열 개수를 하나의 표로 정리하세요.'''
shape_summary = pd.DataFrame([
    {
        "dataset": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
    }
    for name, df in datasets.items()
])

shape_summary
```

```python
'''2. 각 파일의 결측치 개수 확인하세요.'''
missing_summary = pd.concat(
    [df.isna().sum().rename(name) for name, df in datasets.items()],
    axis=1,
).fillna("").astype(str)

missing_summary
```

```python
'''2. 각 파일의 결측치 비율을 확인하세요.'''
missing_rate_summary = pd.concat(
    [(df.isna().mean() * 100).round(2).rename(name) for name, df in datasets.items()],
    axis=1,
).fillna("")

missing_rate_summary
```

```python
'''3. `orders.order_date`를 날짜 타입으로 변환하고 데이터 기간을 확인하세요.'''
orders["order_date"] = pd.to_datetime(orders["order_date"], errors="coerce")
orders.assign(
    order_year=orders["order_date"].dt.year,
    order_month=orders["order_date"].dt.month,
    order_day_name=orders["order_date"].dt.day_name(),
).head()
```

```python
'''4. `order_items`와 `products`를 병합해 카테고리별 주문 금액을 계산하세요.'''
order_items_with_products = order_items.merge(
    products,
    on="product_id",
    how="left",
)

order_items_with_products["line_amount"] = (
    order_items_with_products["quantity"] * order_items_with_products["unit_price"]
)
category_sales = (
    order_items_with_products
    .groupby("category", as_index=False)["line_amount"]
    .sum()
    .sort_values("line_amount", ascending=False)
)

category_sales
```
